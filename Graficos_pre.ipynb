```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px # Import plotly.express to get a list of qualitative colors

# Ensure 'bedtime_order' and 'smoking_habit_order' are defined and 'During_Days of the week bedtime later than usual ' is cleaned
df['Pre_Days of the week bedtime later than usual '] = df['Pre_Days of the week bedtime later than usual '].replace({
    '1 to 2': '1 to 2 days/week',
    '3 to 4': '3 to 4 days/week',
    '5 to 7': '5 to 7 days/week',
    'Never': 'Never'
})

bedtime_order = ['Never', '1 to 2 days/week', '3 to 4 days/week', '5 to 7 days/week']
smoking_habit_order = [
    "I don't smoke",
    "Has decreased considerably",
    "It has decreased but only a little",
    "It has been maintained",
    "Has increased slightly",
    "It has increased quite a bit",
    "Has increased considerably"
]

# Identify the PSQI score column
psqi_col_name = df.columns[32]

# Initialize the subplot figure with 2 rows and 2 columns
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=(
                        "PSQI score by Sun Exposure",
                        "PSQI score by Lifestyle",
                        "PSQI score by Bedtime Later",
                        "PSQI score by Smoking Habit"
                    ))

colors_list = px.colors.qualitative.Plotly # A list of distinct colors
current_color_idx = 0

# Subplot 1: Sun Exposure
sun_exposure_levels = sorted(df['DURING_Sun exposure'].unique(), key=lambda x: ['I don\'t expose myself to the sun', 'Less than 1 time a week', '1 to 2 times a week', '3 to 4 times a week', '5 to 7 times a week', 'Once a month'].index(x) if x in ['I don\'t expose myself to the sun', 'Less than 1 time a week', '1 to 2 times a week', '3 to 4 times a week', '5 to 7 times a week', 'Once a month'] else 99)
for level in sun_exposure_levels:
    fig.add_trace(go.Box(
        y=df[df['DURING_Sun exposure'] == level][psqi_col_name],
        name=f"Sun Exposure: {level}", # Name for legend
        marker_color=colors_list[current_color_idx % len(colors_list)],
        showlegend=True,
        legendgroup='Sun Exposure'
    ), row=1, col=1)
    current_color_idx += 1
fig.update_xaxes(title_text="", row=1, col=1, tickangle=45, categoryorder='array', categoryarray=sun_exposure_levels, showticklabels=False)
fig.update_yaxes(title_text="pre-PSQI score", row=1, col=1)

# Subplot 2: Lifestyle
lifestyle_levels = df['Pre_Lifestyle '].unique()
for level in lifestyle_levels:
    fig.add_trace(go.Box(
        y=df[df['Pre_Lifestyle '] == level][psqi_col_name],
        name=f"Lifestyle: {level}",
        marker_color=colors_list[current_color_idx % len(colors_list)],
        showlegend=True,
        legendgroup='Lifestyle'
    ), row=1, col=2)
    current_color_idx += 1
fig.update_xaxes(title_text="", row=1, col=2, tickangle=45, showticklabels=False)
fig.update_yaxes(title_text="pre-PSQI score", row=1, col=2)

# Subplot 3: Bedtime Later
for level in bedtime_order:
    if level in df['Pre_Days of the week bedtime later than usual '].unique(): # Only add if data exists
        fig.add_trace(go.Box(
            y=df[df['Pre_Days of the week bedtime later than usual '] == level][psqi_col_name],
            name=f"Bedtime: {level}",
            marker_color=colors_list[current_color_idx % len(colors_list)],
            showlegend=True,
            legendgroup='Bedtime Later'
        ), row=2, col=1)
        current_color_idx += 1
fig.update_xaxes(title_text="", row=2, col=1, tickangle=45, categoryorder='array', categoryarray=bedtime_order, showticklabels=False)
fig.update_yaxes(title_text="pre-PSQI score", row=2, col=1)

# Subplot 4: Smoking Habit
for level in smoking_habit_order:
    if level in df['During_smoking habit (Don´t smoke= 1,   decreased considerably= 2,  decreased a little= 3, has been maintained= 4,  increased slightly= 5,'].unique(): # Only add if data exists
        fig.add_trace(go.Box(
            y=df[df['During_smoking habit (Don´t smoke= 1,   decreased considerably= 2,  decreased a little= 3, has been maintained= 4,  increased slightly= 5,'] == level][psqi_col_name],
            name=f"Smoking: {level}",
            marker_color=colors_list[current_color_idx % len(colors_list)],
            showlegend=True,
            legendgroup='Smoking Habit'
        ), row=2, col=2)
        current_color_idx += 1
fig.update_xaxes(title_text="", row=2, col=2, tickangle=45, categoryorder='array', categoryarray=smoking_habit_order, showticklabels=False)
fig.update_yaxes(title_text="pre-PSQI score", row=2, col=2)

# Update overall layout for the figure
fig.update_layout(height=800, width=1200,
                  title_text="Combined Boxplots of PSQI Score by Sun Exposure, Lifestyle, Bedtime Later, and Smoking Habit")

fig.show()
```
